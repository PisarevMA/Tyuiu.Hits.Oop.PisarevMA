<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс ShippingOption в C#, который будет представлять различные опции доставки. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, a также некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу ShippingOption:
    Атрибуты: ID опции доставки (DeliveryOptionld), Название опции доставки (Delivery OptionName), Стоимость доставки (Cost).
    Методы:
        Calculate Cost(): метод для расчета стоимости доставки.
        Estimate Delivery Time(): метод для оценки времени доставки.
        GetDeliveryDetails(): метод для получения деталей опции доставки.
Требования к производным классам:
1. СтандартнаяДоставка (StandardDelivery): Должна содержать дополнительные атрибуты, такие как Среднее время доставки (AverageDeliveryTime). Метод EstimateDeliveryTime() должен быть переопределен для предоставления среднего времени доставки.
2. ЭкспрессДоставка (ExpressDelivery): Должна содержать дополнительные атрибуты, такие как Минимальное время доставки (MinDeliveryTime). Метод CalculateCost() должен быть переопределен для увеличения стоимости доставки в случае необходимости ускорения доставки.
3. Самовывоз (Pickup) (если требуется третий класс): Должна содержать дополнительные атрибуты, такие как Адрес пункта самовывоза (PickupAddress). Метод GetDeliveryDetails() должен быть переопределен для отображения адреса пункта самовывоза.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
public interface ITrackable
{
    string GetTrackingStatus();
}

public interface IInsurable
{
    decimal CalculateInsurance();
}

public class ShippingOption
{
    private int _deliveryOptionId;
    private string _deliveryOptionName;
    private decimal _cost;
    private string _carrierName;
    private int _maxWeight;
    private bool _isFragile;

    public int DeliveryOptionId
    {
        get { return _deliveryOptionId; }
        set { _deliveryOptionId = value; }
    }

    public string DeliveryOptionName
    {
        get { return _deliveryOptionName; }
        set { _deliveryOptionName = value; }
    }

    public decimal Cost
    {
        get { return _cost; }
        set { _cost = value; }
    }

    public string CarrierName
    {
        get { return _carrierName; }
        set { _carrierName = value; }
    }

    public int MaxWeight
    {
        get { return _maxWeight; }
        set { _maxWeight = value; }
    }

    public bool IsFragile
    {
        get { return _isFragile; }
        set { _isFragile = value; }
    }

    public ShippingOption(int id, string name, decimal cost)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = cost;
        CarrierName = "Не указан";
        MaxWeight = 20;
        IsFragile = false;
    }

    public ShippingOption(ShippingOption other)
    {
        DeliveryOptionId = other.DeliveryOptionId;
        DeliveryOptionName = other.DeliveryOptionName;
        Cost = other.Cost;
        CarrierName = other.CarrierName;
        MaxWeight = other.MaxWeight;
        IsFragile = other.IsFragile;
    }

    public virtual decimal CalculateCost()
    {
        return Cost;
    }

    public virtual int EstimateDeliveryTime()
    {
        return 5;
    }

    public virtual string GetDeliveryDetails()
    {
        return $"ID: {DeliveryOptionId}, Название: {DeliveryOptionName}, Стоимость: ${Cost}";
    }

    public virtual bool CanDeliver(int weight)
    {
        return weight <= MaxWeight;
    }

    public virtual string GetCarrierInfo()
    {
        return $"Перевозчик: {CarrierName}, Макс. вес: {MaxWeight} кг, Хрупкий: {IsFragile}";
    }
}

public class StandardDelivery : ShippingOption
{
    private int _averageDeliveryTime;
    private string _deliveryZone;
    private bool _isWeekendDelivery;

    public int AverageDeliveryTime
    {
        get { return _averageDeliveryTime; }
        set { _averageDeliveryTime = value; }
    }

    public string DeliveryZone
    {
        get { return _deliveryZone; }
        set { _deliveryZone = value; }
    }

    public bool IsWeekendDelivery
    {
        get { return _isWeekendDelivery; }
        set { _isWeekendDelivery = value; }
    }

    public StandardDelivery(int id, string name, decimal cost, int avgTime)
        : base(id, name, cost)
    {
        AverageDeliveryTime = avgTime;
        DeliveryZone = "Город";
        IsWeekendDelivery = false;
    }

    public StandardDelivery(StandardDelivery other)
        : base(other)
    {
        AverageDeliveryTime = other.AverageDeliveryTime;
        DeliveryZone = other.DeliveryZone;
        IsWeekendDelivery = other.IsWeekendDelivery;
    }

    public override int EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $"\nЗона доставки: {DeliveryZone}, Доставка в выходные: {IsWeekendDelivery}";
    }

    public string GetStandardInfo()
    {
        return $"Стандартная доставка: {DeliveryOptionName}, срок {AverageDeliveryTime} дней";
    }
}

public abstract class ExpressDeliveryBase : ShippingOption
{
    protected int _minDeliveryTime;
    protected string _courierName;

    public int MinDeliveryTime
    {
        get { return _minDeliveryTime; }
        set { _minDeliveryTime = value; }
    }

    public string CourierName
    {
        get { return _courierName; }
        set { _courierName = value; }
    }

    protected ExpressDeliveryBase(int id, string name, decimal cost, int minTime)
        : base(id, name, cost)
    {
        MinDeliveryTime = minTime;
        CourierName = "Курьер №1";
    }

    protected ExpressDeliveryBase(ExpressDeliveryBase other)
        : base(other)
    {
        MinDeliveryTime = other.MinDeliveryTime;
        CourierName = other.CourierName;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $"\nКурьер: {CourierName}, Минимум дней: {MinDeliveryTime}";
    }
}

public class ExpressDelivery : ExpressDeliveryBase, ITrackable, IInsurable
{
    private const decimal ExpressMultiplier = 1.5m;
    private bool _isSameDay;

    public bool IsSameDay
    {
        get { return _isSameDay; }
        set { _isSameDay = value; }
    }

    public ExpressDelivery(int id, string name, decimal cost, int minTime)
        : base(id, name, cost, minTime)
    {
        IsSameDay = false;
    }

    public ExpressDelivery(ExpressDelivery other)
        : base(other)
    {
        IsSameDay = other.IsSameDay;
    }

    public override decimal CalculateCost()
    {
        return Cost * ExpressMultiplier;
    }

    public override int EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $"\nДоставка в тот же день: {IsSameDay}";
    }

    public string GetExpressInfo()
    {
        return $"Экспресс доставка: {DeliveryOptionName}, минимум {MinDeliveryTime} день(дней)";
    }

    public string GetTrackingStatus()
    {
        return $"Экспресс доставка \"{DeliveryOptionName}\": посылка в пути";
    }

    public decimal CalculateInsurance()
    {
        return CalculateCost() * 0.1m;
    }
}

public class Pickup : ShippingOption
{
    private string _pickupAddress;
    private string _workingHours;
    private int _storageDays;

    public string PickupAddress
    {
        get { return _pickupAddress; }
        set { _pickupAddress = value; }
    }

    public string WorkingHours
    {
        get { return _workingHours; }
        set { _workingHours = value; }
    }

    public int StorageDays
    {
        get { return _storageDays; }
        set { _storageDays = value; }
    }

    public Pickup(int id, string name, decimal cost, string address)
        : base(id, name, cost)
    {
        PickupAddress = address;
        WorkingHours = "09:00 - 18:00";
        StorageDays = 7;
    }

    public Pickup(Pickup other)
        : base(other)
    {
        PickupAddress = other.PickupAddress;
        WorkingHours = other.WorkingHours;
        StorageDays = other.StorageDays;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $"\nАдрес самовывоза: {PickupAddress}";
    }

    public string GetPickupInfo()
    {
        return $"Самовывоз: {PickupAddress}, часы работы: {WorkingHours}, хранение: {StorageDays} дней";
    }
}

public class DeliveryService
{
    private string _serviceName;
    private List<ShippingOption> _options;

    public string ServiceName
    {
        get { return _serviceName; }
        set { _serviceName = value; }
    }

    public List<ShippingOption> Options
    {
        get { return _options; }
        set { _options = value; }
    }

    public DeliveryService(string serviceName)
    {
        ServiceName = serviceName;
        Options = new List<ShippingOption>();
    }

    public DeliveryService(string serviceName, List<ShippingOption> options)
    {
        ServiceName = serviceName;
        Options = options;
    }

    public void AddOption(ShippingOption option)
    {
        Options.Add(option);
    }

    public ShippingOption FindCheapest()
    {
        ShippingOption cheapest = Options[0];
        foreach (var option in Options)
        {
            if (option.CalculateCost() < cheapest.CalculateCost())
            {
                cheapest = option;
            }
        }
        Console.WriteLine($"Самая дешёвая опция в \"{ServiceName}\": {cheapest.DeliveryOptionName}");
        return cheapest;
    }

    public ShippingOption FindFastest()
    {
        ShippingOption fastest = Options[0];
        foreach (var option in Options)
        {
            if (option.EstimateDeliveryTime() < fastest.EstimateDeliveryTime())
            {
                fastest = option;
            }
        }
        Console.WriteLine($"Самая быстрая опция в \"{ServiceName}\": {fastest.DeliveryOptionName}");
        return fastest;
    }
}

var options = new List<ShippingOption>
{
    new StandardDelivery(1, "Стандартная доставка", 10.00m, 7),
    new ExpressDelivery(2, "Экспресс доставка", 15.00m, 1),
    new Pickup(3, "Самовывоз", 0.00m, "ул. Ленина, д. 5")
};

DeliveryService service = new DeliveryService("ТИУ почта", options);

foreach (var option in service.Options)
{
    Console.WriteLine($"{option.DeliveryOptionName}:");
    Console.WriteLine($"Стоимость: ${option.CalculateCost()}");
    Console.WriteLine($"Время доставки (дней): {option.EstimateDeliveryTime()}");
    Console.WriteLine(option.GetDeliveryDetails());
    Console.WriteLine(option.GetCarrierInfo());
    Console.WriteLine();
}

service.FindCheapest();
service.FindFastest();
```
